```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("/Users/aftaabmulla/Capstone/updated_data.csv", sep=",")
```

```python
print("Duplicates before:", data.duplicated().sum())

data.drop_duplicates(inplace=True)

print("Duplicates after:", data.duplicated().sum())
```

```python
missing = data.isnull().sum()
missing
```

```python
# Remove the unnecessary column
if "Unnamed: 9" in data.columns:
    data.drop(columns=["Unnamed: 9"], inplace=True)

# Check missing values before removing rows
print("Missing values before:")
print(data.isnull().sum())

# Remove rows containing any missing values
data.dropna(inplace=True)

# Reset the index
data.reset_index(drop=True, inplace=True)

# Check dataset information
print("\nDataset shape after removing missing rows:", data.shape)

print("\nMissing values after:")
print(data.isnull().sum())
```

```python
import seaborn as sns

plt.figure(figsize=(6,4))

sns.countplot(x="level", data=data)

plt.title("State vs Central Schemes")

plt.show()
```

```python
plt.figure(figsize=(12,6))

data["schemeCategory"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Scheme Categories")

plt.ylabel("Count")

plt.show()
```

```python
from collections import Counter

tags = ",".join(data["tags"])

tags = tags.split(",")

counter = Counter(tags)

top = counter.most_common(15)

top_df = pd.DataFrame(top, columns=["Tag","Count"])

plt.figure(figsize=(12,6))

sns.barplot(data=top_df,x="Count",y="Tag")

plt.title("Top Tags")

plt.show()
```
