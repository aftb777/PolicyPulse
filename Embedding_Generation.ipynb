```python
import pandas as pd

from sentence_transformers import SentenceTransformer
```

```python
df = pd.read_csv("PolicyPulse_Cleaned.csv")

df.head()
```

```python
# Create Combined Text
df["combined_text"] = (
    df["scheme_name"].fillna("") + " " +
    df["details"].fillna("") + " " +
    df["benefits"].fillna("") + " " +
    df["eligibility"].fillna("") + " " +
    df["application"].fillna("") + " " +
    df["documents"].fillna("") + " " +
    df["tags"].fillna("")
)
```

```python
df.head()
```

```python
# Download Embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
# Generate Embeddings
embeddings = model.encode(
    df["combined_text"].tolist(),
    show_progress_bar=True,
    convert_to_numpy=True
)
```

```python
print(embeddings.shape)
```

```python
# Save the folder in NP format 
import numpy as np

np.save("scheme_embeddings.npy", embeddings)
```
